## Merge JSONs of Same Publication

```python
import json
import re


def merge_annotation_files(file_paths, output_file):

    # Sort files by the number before "of"
    file_paths = sorted(
        file_paths,
        key=lambda x: int(re.search(r'_(\d+)of\d+', x).group(1))
    )

    merged_data = {
        'classes': [],
        'annotations': []
    }

    for i, file_path in enumerate(file_paths):

        with open(file_path, 'r', encoding='utf-8') as f:
            data = json.load(f)

        # Classes are the same, so take them only from the first file
        if i == 0:
            merged_data['classes'] = data['classes']

        # Add annotations in file order
        merged_data['annotations'].extend(data['annotations'])

    # Save merged JSON
    with open(output_file, 'w', encoding='utf-8') as f:
        json.dump(merged_data, f, indent=2, ensure_ascii=False)

    return merged_data
```
